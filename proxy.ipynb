```python
class Animal:
    def speak():
        pass

class Cat(Animal):
    def speak(self):
        print('meow')

kitty = Cat()
kitty.speak()
```

```python
#인터페이가 같아야하므로 상속을 받아도 됨
class CatProxy(Animal):
    def __init__(self,cat:Cat):
        self.cat = cat

    #위 cat의 인터페이스와 proxy의 인터페스가 같아야함
    def speak(self):
        print("before speak")
        self.cat.speak()
        print("after speak")

kitty_proxy = CatProxy(kitty)
kitty_proxy.speak()

    
```
