```python
class Component:
    def fn(self):
        pass

class Leaf(Component):
    def fn(self):
        print('leaf')

class Composite(Component):
    def __init__(self):
        self.components=[]
    def add(self, component:Component):
        self.components.append(component)

    def fn(self):
        print('composite')
        for component in self.components:
            component.fn()
```

```python
compost1 = Composite()
compost1.add(Leaf())
compost1.add(Leaf())
#self.components에 leaf,leaf
compst0 = Composite()
compst0.add(Leaf())
compst0.add(compost1)
#self.components에 leaf compost1



compst0.fn()
```
