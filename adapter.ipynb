```python
class Animal:
    def walk(self):
        pass

class Cat(Animal):
    def walk(self):
        print("cat walk")

class Dog(Animal):
    def walk(self):
        print("dog walk")

def makeWalk(animal:Animal):
    animal.walk()

cat = Cat()
makeWalk(cat)

dog = Dog()
makeWalk(dog)
```

```python
class Fish:
    def swim(self):
        print("fish swimming")

fish = Fish()
#에러 발생
#makeWalk(fish)
```

```python
class FishAdapter(Animal):
    def __init__(self, fish:Fish):
        self.fish = fish

    def walk(self):
        self.fish.swim()

fish_adapter = FishAdapter(fish)

makeWalk(fish_adapter)
```
